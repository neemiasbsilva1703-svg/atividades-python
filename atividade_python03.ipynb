```python
#questão-12
produtos = {"acuçar": 12, "arroz": 5, "café": 9, "leite": 7,}
produto = input("Digite o nome do produto: ")
vendida = int(input("digite a quantidade de produtos: "))
if produto in produtos:
  if produtos[produto] >= vendida:
    produtos[produto] -= vendida
    print(f"venda realizada!! o produto`{produto} tem {produtos[produto]} de estoque")
  else:
    print(f"Estoque insuficiente! Disponível: {produtos[produto]} unidades.")
else:
  print("produto não cadastradO!!")
for produto, quantidade in produtos.items():
    print(f"{produto}: {quantidade}")
```

```python
#questão-15
estoque = [{"codigo": 101, "nome": "Arroz", "quantidade": 100, "preco": 25.55},
    {"codigo": 102, "nome": "Feijao", "quantidade": 100, "preco": 8.50},
    {"codigo": 103, "nome": "Acucar", "quantidade": 100, "preco": 5.00},
    {"codigo": 104, "nome": "leite", "quantidade": 100, "preco": 7.00},
    {"codigo": 105, "nome": "Cafe", "quantidade": 100, "preco": 15.75}
]
for item in estoque:
 print(f"Código: {item['codigo']} | Nome: {item['nome']} | "
          f"Quantidade: {item['quantidade']} | Preço: R$ {item['preco']:.2f}")
valor_total = 0
for item in estoque:
 valor_total += item["quantidade"] * item["preco"]

 maior_item = estoque[0]
maior_valor = maior_item["quantidade"] * maior_item["preco"]

for item in estoque[1:]:
    valor_item = item["quantidade"] * item["preco"]
    if valor_item > maior_valor:
        maior_valor = valor_item
        maior_item = item
print(f"Item de maior valor em estoque: {maior_item['nome']} "
      f"(R$ {maior_valor:.2f})")

print(f"O valor valor total do estoque é de: R$ {valor_total:.2f}")
resumo = (len(estoque), valor_total)

print(f"\nResumo (quantidade de itens, valor total): {resumo}")
```
