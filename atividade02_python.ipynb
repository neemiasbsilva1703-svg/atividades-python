```python
#questão-15
saldo = 200
qtd_entradas = 0
qtd_retiradas_ok = 0
qtd_retiradas_recusadas = 0
total_entrou = 0
total_saiu = 0

codigo = int(input("Digite o código (1=entrada, 2=retirada, 0=sair): "))

while codigo != 0:
    if codigo == 1:
        valor = int(input("Digite o valor: "))
        saldo = saldo + valor
        qtd_entradas = qtd_entradas + 1
        total_entrou = total_entrou + valor
    elif codigo == 2:
        valor = int(input("Digite o valor: "))
        if valor <= saldo:
            saldo = saldo - valor
            qtd_retiradas_ok = qtd_retiradas_ok + 1
            total_saiu = total_saiu + valor
        else:
            print("Saldo insuficiente")
            qtd_retiradas_recusadas = qtd_retiradas_recusadas + 1

    codigo = int(input("Digite o código (1=entrada, 2=retirada, 0=sair): "))

print(f"Saldo final: R$ {saldo:.2f}")
print(f"Quantidade de entradas: {qtd_entradas}")
print(f"Quantidade de retiradas realizadas: {qtd_retiradas_ok}")
print(f"Quantidade de retiradas recusadas: {qtd_retiradas_recusadas}")
print(f"Total que entrou: R$ {total_entrou:.2f}")
print(f"Total que saiu: R$ {total_saiu:.2f}")
```
